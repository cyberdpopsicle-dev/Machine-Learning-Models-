```python
import pickle
import numpy as np
from sklearn.datasets import load_digits
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
# 1. Load data
digits = load_digits()
X_train, X_test, y_train, y_test = train_test_split(
    digits.data, digits.target, test_size=0.02, random_state=42)
# 2. Train the Base Model
base_model = LogisticRegression(max_iter=1000)
base_model.fit(X_train, y_train)
# Check initial accuracy
initial_acc = base_model.score(X_test, y_test)
print(f"Base Model trained successfully. Test Accuracy: {initial_acc * 100:.2f}%")
# 3. Save to Pickle
with open("base_model.pkl", "wb") as f:
    pickle.dump(base_model, f)
print("Base model saved as 'base_model.pkl'\n")
```

```python
def absolute_unlearn(model_path):
    "Loads a pickled model and completely wipes its learned weights without knowing anything about the original training process. "
    # Load the model blindly
    with open(model_path, "rb") as f:
        loaded_model = pickle.load(f)
    print(" Unlearning Process Started ")
    # Check if it's a scikit-learn linear/logistic model
    if hasattr(loaded_model, "coef_"):
        # Scramble or zero out the weights (coefficients)
        # We use random noise here so it completely forgets the patterns
        loaded_model.coef_ = np.random.normal(0, 1, loaded_model.coef_.shape)
        # Scramble the intercepts as well
        if hasattr(loaded_model, "intercept_"):
            loaded_model.intercept_ = np.random.normal(0, 1, loaded_model.intercept_.shape)
        print("Success: Model weights and intercepts have been entirely randomized!")
    else:
        print("Error: Model type not explicitly supported for weight-shattering.")     
    return loaded_model
# Run the unlearning model
unlearnt_model = absolute_unlearn("base_model.pkl")
```

```python
# Evaluate the unlearnt model
unlearnt_acc = unlearnt_model.score(X_test, y_test)
print(f"\nPost-Unlearning Test Accuracy: {unlearnt_acc * 100:.2f}%")
# A purely random guess on 10 digits should be roughly 10%
print(f"Expected random guess accuracy: ~10.00%")
```
